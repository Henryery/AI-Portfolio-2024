```python
# DEBUGGING STARTER PACK
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Load/simplify data (no missing values)
data = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
X = data[['Pclass', 'Age', 'Fare', 'Sex']].copy()
X['Sex'] = X['Sex'].map({'male': 0, 'female': 1})  # Convert gender to numbers
X = X.fillna(0)  # Fill missing ages/fares
y = data['Survived']

# Verify data
print("Missing values:", X.isnull().sum())
print("First 5 rows:\n", X.head())
```

```python
# SAFE GRIDSEARCH (copy-paste exactly)
params = {
    'n_estimators': [50, 100],  # Fewer options = faster
    'max_depth': [3, 5],        # Skip 'None' for now
    'min_samples_split': [2, 5]
}

grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid=params,
    cv=3,  # Fewer folds = faster
    scoring='accuracy'
)
grid.fit(X, y)  # Use the X/y we just cleaned
print("Best params:", grid.best_params_)
```

```python
# FINAL MODEL (using best params)
best_model = RandomForestClassifier(
    n_estimators=100,  # ← Your 'best_params' values here
    max_depth=5,
    min_samples_split=2,
    random_state=42
)
best_model.fit(X, y)
print(f"Final Accuracy: {best_model.score(X, y):.2f}")
```
